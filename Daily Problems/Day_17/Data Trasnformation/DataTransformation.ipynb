```python
def shows(i):
        if i<=0:
            return 'ERROR!!!!!'
        elif i < 50:
            return 'Low'
        elif i>=50 and i<80:
            return 'Intermediate'
        elif i >=80 and i <100:
            return 'High'
        else:
            return ' Extraorinary'
```

```python
df2['SHOWS']=df2['SHOWS'].apply(shows)
```
