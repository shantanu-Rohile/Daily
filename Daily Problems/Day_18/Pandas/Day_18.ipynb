```python
# 7. Merge two Pandas DataFrames based on a common column.
data1 = {
  'name' : ['jvm','python','sql'],
  'size' : [100,2500,800]
}

data1 = pd.DataFrame(data1)

data2 = {
  'name' : ['jvm','python','sql'],
  'id' : [1,2,3]
}

data2 = pd.DataFrame(data2)
```

```python
merged_df = pd.merge(data1,data2,on='name',how='inner')
```

```python
print(merged_df)
```

```python
# 8. Extract rows from a Pandas DataFrame where a specific column's values are in a given NumPy array.

idx = [0,1]

for i in idx:
  if i in merged_df.iloc[:,0]:
    print(merged_df.iloc[i,0])
```

```python
# 9. Perform element-wise addition of a NumPy array and a Pandas DataFrame column.


data = {'Name': ['Teodosija', 'Sutton', 'Taneli', 'David', 'Ross'],
        'Age': [25, 30, 22, 35, 28],
        'Salary': [50000, 60000, 45000, 70000, 55000]}

add = np.array([10000,10000,10000,10000,10000])
df_add = pd.DataFrame(data)

df_add['update_Salary'] = df_add['Salary'] + add
```

```python
df_add['update_Salary']
```
