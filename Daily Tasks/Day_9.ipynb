```python
# 1.python set

# Create a set called colors with the values
colors = {"red", "green", "blue"}
# Print the set
print("orignal : ",colors)
# Add "yellow" to the set using add()
colors.add("yellow")
print("After adding yellow : ",colors)
# Remove "green" from the set using discard()
colors.remove("green")
print("After removing green : ",colors)
# Print the number of items using len()
print("LENGTH : ",len(colors))
```

```python
# 2.python dictionaries

# Create a dictionary called car with the keys "brand", "model", "year" and values "Ford", "Mustang", 2024

car = {"brand" : "Ford", "model" : "Mustang", "year" : 2024}

# Print the value of the "model" key

print(car["model"])

# Add a new key "color" with the value "red"

car["color"] = "red"

# Remove the "brand" key using pop()

car.pop("brand")

# Print the dictionary

print(car)
```

```python
# 4.while loop
# Create a variable i with the value 0
i = 0
# Write a while loop that runs as long as i is less than 6
# Inside the loop: increment i by 1
# If i equals 3, use continue to skip that iteration
# Print i
while i < 6 :
  i = i + 1
  if i == 3 :
    continue

  print(i)


```

```python
# 5.for loop

# Create a list called fruits with: "apple", "banana", "cherry"
fruits = ["apple", "banana", "cherry"]
# Write a for loop that prints each item in fruits
# Use break to stop the loop when the item is "banana"
for i in fruits:
  if i == "banana":
    break
  print(i)

```

```python
from os import name
# 7.what are objects?
# Create a class called Person
# Add an __init__ method that takes name and age as parameters
# Add a method called greet that prints "Hello, my name is " followed by the name
# Create an object p1 of the class with name "John" and age 36
# Call the greet method on p1

# objects are instance of class

class Person :
  def __init__(self,name,age):
    self.name = name
    self.age = age
  def greet(self):
    print("Hello Mr./Mrs.", self.name,"!")

p1 = Person("Shantanu",22)
print(p1.age)
p1.greet()
```

```python
# 8.what are classes?

# classes are blueprint that we use to add real world ideas into the code

# Create a class Student with an __init__ that takes name and grade, and stores them as properties
# Create an object s1 with name "Anna" and grade "A"
# Print the grade of s1
# Change the grade of s1 to "B"
# Print the updated grade

class Student:
  def __init__(self,name,grade):
    self.name=name
    self.grade= grade

s1 = Student("Anna","A")
s1.grade = "B"
print(s1.grade)
```
