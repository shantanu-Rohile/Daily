```python
# 1.Given two lists, l1 and l2, write a program to create a third list l3 
# by picking an odd-index element from the list l1 and even index elements from the list l2.
eleven = [3, 6, 9, 12, 15, 18, 21]
twelve = [4, 8, 12, 16, 20, 24, 28]
res=[]
for i in range(len(eleven)):
    if i%2 != 0 :
        res.append(eleven[i])
    else:
        res.append(twelve[i])

print(res)
    
```

```python
# 3.Count the occurrence of each element from a list
sample_list = [11, 45, 8, 11, 23, 45, 23, 45, 89]

frequency ={}

for i in sample_list:
    if i in frequency:
        frequency.update({i:frequency.get(i)+1})
    else:
        frequency.update({i:1})

print(frequency)
```

```python
# 4. Create a Python set such that it shows the element from both lists in a pair
first_list = [2, 3, 4, 5, 6, 7, 8]
second_list = [4, 9, 16, 25, 36, 49, 64]

first_list = set(first_list)
second_list = set(second_list)

print(first_list.union(second_list))
```

```python
# 5.Find the intersection (common) of two sets and remove those elements from the first set
first_set = {23, 42, 65, 57, 78, 83, 29}
second_set = {57, 83, 29, 67, 73, 43, 48}

print(first_set-first_set.intersection(second_set))
```

```python
# 6. Remove duplicates from a list and create a tuple and find the minimum and maximum number
sample_list = [87, 45, 41, 65, 94, 41, 99, 94]

sample_list = set(sample_list)

sample_list = tuple(sample_list)

min = 10000
max = -10000


for i in sample_list:
    if i>=max:
        max =i 
    if i<=min:
        min=i

print("Minimum :",min)

print("Maximum :",max)
```
