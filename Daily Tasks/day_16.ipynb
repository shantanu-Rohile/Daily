# 2.What is the use of break statement?

- Break statement is used to terminate the loop when certain conditions are fullfield

```python
n = 10

c = 7

for i in range(n):
    if i==c:
        break
    print(i,end=" ")
```

# 4. What is a generator in Python?

- Generators allows us to create our own iterators in simple way without using class

```python
def generator():
    yield 10

    yield 11

    yield 12

    yield 13

    yield 14


g = generator()

print(g.__next__())
print(g.__next__())
print(g.__next__())
# for i in g:
#     print(i)
```
